# list containing three dictionaries.

```python
predictions = [
    {"label": "cat", "confidence": 0.91},
    {"label": "dog", "confidence": 0.23},
    {"label": "bird", "confidence": 0.87},
]
```

access the data

```python
print(predictions[0])
# {'label': 'cat', 'confidence': 0.91}

print(predictions[0]["label"])
# cat

print(predictions[1]["confidence"])
# 0.23
```

# list_name[index]["dictionary_key"]

```python
predictions = [
    {"label": "cat", "confidence": 0.91},
    {"label": "dog", "confidence": 0.23},
    {"label": "bird", "confidence": 0.87},
    {"label": "horse", "confidence": 0.76},
]
```

```python
high_predictions=[]
for prediction in predictions:
    if prediction["confidence"]>0.8:
        high_predictions.append(prediction)
        
```

```python
print(high_predictions)
```

```python
high_predictions=[
    prediction
    for prediction in predictions
    if prediction["confidence"]>0.8
]
```

```python
print(high_predictions)
```

```python
for prediction in predictions:
    print(prediction["label"])
```
