# *args and **kwargs
*args collects extra "positional arguments" into a tuple
**kwargs collects extra "keyword arguments" into a dict

```python
def example(*args,**kwargs):
    print(args)
    print(kwargs)
    
```

```python
example(1,2,model="gemini",temp=0.5)
```

# Functions that work with dictionaries and lists

```python
def get_high_confidence(predictions:list[dict])-> list[dict]:
    return [
        prediction
        for prediction in predictions
        if prediction["confidence"]>0.8
    ]
```
